## Bài 2. Tạo các Tensor thường dùng 
Thực hành tạo Tensor ngẫu nhiên, Tensor toàn số 0, toàn số 1 và ma trận đơn vị. 

```python
import torch

# 1. Khởi tạo các ma trận đặc biệt cơ bản (ví dụ với kích thước 3x3)
zeros_tensor = torch.zeros(3, 3)
print("1. Ma trận toàn số 0:\n", zeros_tensor)

ones_tensor = torch.ones(3, 3)
print("\n2. Ma trận toàn số 1:\n", ones_tensor)

eye_tensor = torch.eye(3)
print("\n3. Ma trận đơn vị (đường chéo bằng 1):\n", eye_tensor)

# 2. Khởi tạo ma trận ngẫu nhiên
# Phân bố đều (các giá trị nằm trong khoảng từ 0 đến 1)
rand_tensor = torch.rand(3, 3)
print("\n4. Ma trận ngẫu nhiên đều (torch.rand):\n", rand_tensor)

# Phân bố chuẩn (xoay quanh giá trị 0, có cả số âm và dương)
randn_tensor = torch.randn(3, 3)
print("\n5. Ma trận ngẫu nhiên chuẩn (torch.randn):\n", randn_tensor)

# 3. Code yêu cầu: Tạo ma trận 5x5 toàn số 10
tensor_10_v1 = torch.ones(5, 5) * 10
print("\n6. Ma trận 5x5 toàn số 10 (Cách dùng nhân):\n", tensor_10_v1)
```

- torch.randn(): Sinh ra các giá trị ngẫu nhiên theo phân bố chuẩn (normal/Gaussian distribution), có giá trị trung bình bằng 0 và độ lệch chuẩn bằng 1 (nên kết quả có thể chứa cả số âm và số dương).

## Bài 6. Phép toán và nhân ma trận 
Thực hiện các phép toán phần tử và phép nhân ma trận bằng toán tử @. 

```python
import torch

a = torch.randn(2, 3)
b = torch.randn(2, 3)
print("matrix A:")
print(a)
print("matrix B:")
print(b)

# Tính toán thử để xem hình thù kết quả
tich_tung_phan_tu = a * b
tich_ma_tran = a @ b.T
print("tich tung phan:") 
print(tich_tung_phan_tu)
print("tich ma tran:")
print(tich_ma_tran)

print("shape của a@b.T: ", tich_ma_tran.shape)
```

- Phép `a * b` (Nhân element-wise): Là lấy từng phần tử ở cùng vị trí của 2 ma trận nhân trực tiếp với nhau. Yêu cầu hai ma trận (a và b) phải có kích thước y hệt nhau (ở đây cùng là 2x3). Kết quả trả ra một ma trận vẫn giữ nguyên kích thước 2x3.
- Phép `a @ b.T` (Nhân ma trận - Matrix multiplication): Là phép nhân đại số tuyến tính (nhân hàng của ma trận này với cột của ma trận kia). 
  Vì luật nhân ma trận yêu cầu số cột của ma trận đầu phải bằng số hàng của ma trận sau, nên ta không thể nhân a (2x3) với b (2x3) được. Do đó, phải dùng `.T` (Transpose) để đảo ma trận b thành kích thước 3x2. Lúc này (2x3) nhân (3x2) mới hợp lệ.

## Bài 7. Các hàm thống kê trên Tensor 
Tính trung bình, độ lệch chuẩn và tổng tích lũy theo từng cột.

```python
import torch 
 
a = torch.randn(3, 4) 

print(a) 
print("\nMean:") 
print(torch.mean(a, dim=0)) 
 
print("\nStandard deviation:") 
print(torch.std(a, dim=0)) 
 
print("\nCumulative sum:") 
print(torch.cumsum(a, dim=0)) 
```

- Khi dùng `dim=1`: Phép tính được thực hiện ngang qua các HÀNG. Ma trận có 3 hàng nên kết quả sẽ trả về 3 con số (mỗi số là trung bình của một hàng). Nó tính toán bằng cách "ép" các cột lại với nhau.

## Bài 8. Autograd cơ bản 
Tạo biến có requires_grad=True và dùng backward() để tính đạo hàm tự động. 

```python
import torch 
 
x = torch.tensor( 
    3.6, 
    requires_grad=True 
) 
 
y = x * x 
 
y.backward() 
 
print("x =", x) 
print("y =", y) 
print("x.grad =", x.grad) 

x_new = torch.tensor(3.6, requires_grad=True)
y_new = 3 * x_new**2 + 2 * x_new + 1
y_new.backward()
print("x =", x_new) 
print("y =", y_new) 
print("- Kết quả đạo hàm tự động (x_new.grad):", x_new.grad.item())
```

- Tính bằng tay theo Toán Giải tích: y = x^2  => Đạo hàm y' = 2*x.
- Thay x = 3.6 vào, ta có kết quả: y' = 2 * 3.6 = 7.2.
- So sánh: Kết quả tính tay (7.2) trùng khớp hoàn toàn với kết quả x.grad (7.2000) mà PyTorch tự động tính toán từ hàm y.backward().

## Bài 9. Gradient Descent tìm hệ số đa thức 
Sinh dữ liệu từ đa thức y = x² + 2x + 3, sau đó dùng NAdam và autograd để tìm lại ba hệ số.

```python
import numpy as np
import torch

polynomial = np.poly1d([1, 2, 3])
N = 20
X = np.random.randn(N, 1) * 5
Y = polynomial(X)
XX = np.hstack([
    X * X,
    X,
    np.ones_like(X)
])

w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)


optimizer = torch.optim.NAdam([w], lr=0.01)

print("--- QUÁ TRÌNH HỌC CỦA MÁY TÍNH ---")
print("1. Ba hệ số đoán mò ban đầu (ngẫu nhiên):")
print(w.detach().numpy())

for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred)) # Tính sai số MSE
    mse.backward()
    optimizer.step()

print("\n2. Ba hệ số mà máy tính TỰ TÌM ĐƯỢC sau 1000 vòng học:")
print(w.detach().numpy())
```

Nhận xét: Ba hệ số mà AI học được xấp xỉ rất sát với đáp án [1.0, 2.0, 3.0] của hàm đa thức gốc ban đầu. Điều này chứng tỏ thuật toán Gradient Descent đã hoạt động đúng, giúp máy tính từ việc đoán mò ngẫu nhiên ban đầu tiến dần đến được phương trình chính xác nhờ tối ưu hóa sai số.

Giải thích:
1. optimizer.zero_grad(): 
   - Vai trò: Dọn sạch (reset về 0) túi chứa đạo hàm (gradient) của vòng lặp trước đó. 
   - Lý do: PyTorch có cơ chế mặc định là "cộng dồn" đạo hàm. Nếu không dọn sạch đi thì đạo hàm của vòng mới sẽ cộng dính với vòng cũ gây sai số trầm trọng.
   
2. mse.backward(): 
   - Vai trò: Kích hoạt chế độ tính đạo hàm tự động (Autograd). 
   - Lý do: Nó sẽ tính đạo hàm của hàm sai số (mse) theo biến `w`. Ý nghĩa thực tế là nó tìm ra hướng đi (la bàn) để giúp giảm cái sai số này xuống.
   
3. optimizer.step(): 
   - Vai trò: Bước đi cập nhật trọng số. 
   - Lý do: Dựa vào hướng đi đã tính được ở hàm backward, thuật toán optimizer (NAdam) sẽ tiến hành điều chỉnh biến `w` một chút xíu. Cứ lặp lại nhiều lần như vậy, `w` sẽ ngày càng xấp xỉ đáp án đúng.

## Bài 10. Giải hệ phương trình bằng Autograd 
Tối thiểu hóa tổng sai số bình phương để tìm A, B, C, D thỏa bốn phương trình. 

```python
import random
import torch
print("--- TIẾN HÀNH GIẢI HỆ PHƯƠNG TRÌNH ---")
# 1. Khởi tạo 4 biến A, B, C, D ngẫu nhiên (chính là 4 ẩn số cần đi tìm)
A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)
EPOCHS = 2000
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)
# 2. Cho máy tính học (thử nghiệm và sửa sai)
for _ in range(EPOCHS):
    # Đặt 4 phương trình bằng 0 (chuyển vế đổi dấu)
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2
    
    # Sai số bình phương 
    sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4
    
    optimizer.zero_grad()
    sqerr.backward()      
    optimizer.step()     

A_val = round(A.item(), 4)
B_val = round(B.item(), 4)
C_val = round(C.item(), 4)
D_val = round(D.item(), 4)
print("Kết quả nghiệm tìm được:")
print(f"A = {A_val}, B = {B_val}, C = {C_val}, D = {D_val}")
print("\n" + "="*40)
print("--- TRẢ LỜI YÊU CẦU BÀI 10 ---")
print("="*40)

print("1. Kiểm tra lại nghiệm (thử lại vào phương trình):")
print(f"- Phương trình 1 (A + B = 9): {A_val} + {B_val} = {round(A_val + B_val, 4)}")
print(f"- Phương trình 2 (C - D = 1): {C_val} - {D_val} = {round(C_val - D_val, 4)}")
print(f"- Phương trình 3 (A + C = 8): {A_val} + {C_val} = {round(A_val + C_val, 4)}")
print(f"- Phương trình 4 (B - D = 2): {B_val} - {D_val} = {round(B_val - D_val, 4)}")
print("=> Nhận xét: Máy tính đã tìm ra được bộ nghiệm thỏa mãn 100% cả 4 phương trình trên.")
```

Theo Toán Đại số, hệ phương trình này tuy có 4 phương trình nhưng thực chất phương trình số (4) bị phụ thuộc tuyến tính vào 3 phương trình đầu. 
Cụ thể, ta lấy (PT1) - (PT3) + (PT2): 
(A + B) - (A + C) + (C - D) = 9 - 8 + 1 
=> B - D = 2 (Đúng y hệt phương trình số 4).
Vậy hệ thống thực chất chỉ có 3 phương trình độc lập, nhưng lại có tới 4 biến số (A, B, C, D). Quy tắc Toán học chỉ ra rằng: "Hệ có số ẩn nhiều hơn số phương trình độc lập thì hệ đó có VÔ SỐ NGHIỆM".
Vì hệ có vô số nghiệm, nên mỗi lần bạn chạy lại đoạn code trên (với số khởi tạo random ban đầu khác nhau), AI sẽ "dò" ra một bộ nghiệm A,B,C,D hoàn toàn mới nhưng vẫn luôn luôn đúng!

## Bài 11. Viết chương trình PyTorch để thực hiện các yêu cầu sau: 
• Tạo Tensor X có 24 phần tử và reshape lần lượt thành 4 × 6, 2 × 12 và 2 × 3 × 4. 

• Tạo hai Tensor A, B kích thước 3 × 4; tính A+B, A*B, A/B, trung bình theo cột và độ lệch chuẩn theo cột. 

• Cho y = 2x² + 5x + 3. Dùng autograd tính dy/dx tại x = 2 và kiểm tra bằng phép tính thủ công. 

• Thay đa thức trong Bài 9 bằng y = 2x² - 3x + 5; huấn luyện và cho biết các hệ số tìm được. 

```python
import torch
import numpy as np
```

```python
print("\n--- Ý 1: TẠO TENSOR VÀ RESHAPE ---")
# Tạo mảng chứa 24 con số từ 0 đến 23
X = torch.arange(24) 
print("Tensor gốc (24 phần tử):\n", X)
# Reshape
X_4x6 = X.reshape(4, 6)
print("\n- Reshape thành 4x6:\n", X_4x6)
X_2x12 = X.reshape(2, 12)
print("\n- Reshape thành 2x12:\n", X_2x12)
X_2x3x4 = X.reshape(2, 3, 4)
print("\n- Reshape thành 2x3x4:\n", X_2x3x4)
```

```python
print("\n\n--- Ý 2: CÁC PHÉP TOÁN TRÊN TENSOR ---")
A = torch.randn(3, 4)
B = torch.randn(3, 4)
print("Tensor A:\n", A)
print("Tensor B:\n", B)
print("\n- Phép cộng A + B:\n", A + B)
print("- Phép nhân từng phần tử A * B:\n", A * B)
print("- Phép chia từng phần tử A / B:\n", A / B)
# Tính trung bình và độ lệch chuẩn theo cột (dim=0) cho Tensor A
print("\n- Trung bình theo cột của A (Mean dim=0):\n", torch.mean(A, dim=0))
print("- Độ lệch chuẩn theo cột của A (Std dim=0):\n", torch.std(A, dim=0))
```

```python
print("\n\n--- Ý 3: TÍNH ĐẠO HÀM BẰNG AUTOGRAD ---")
# Đặt giá trị x = 2.0 (nhớ dùng số thập phân để PyTorch hiểu là float)
x_val = torch.tensor(2.0, requires_grad=True)
y_val = 2 * x_val**2 + 5 * x_val + 3
# Tính đạo hàm tự động
y_val.backward()
print(f"1. Đạo hàm tự động (Autograd) tại x = 2 là: {x_val.grad.item()}")
# Kiểm tra thủ công
"""
- Tính bằng tay: y' = 4x + 5
- Thay x = 2 vào ta có: y' = 4(2) + 5 = 13
"""
print(f"2. Kiểm tra thủ công: y' = 4*2 + 5 = 13. (Hoàn toàn trùng khớp!)")
```

```python
print("\n\n--- Ý 4: AI TÌM HỆ SỐ ĐA THỨC ---")
# Phương trình gốc là: y = 2x^2 - 3x + 5 (Đáp án chuẩn là 2, -3, 5)
poly_bai11 = np.poly1d([2, -3, 5]) 
N = 20
X_data = np.random.randn(N, 1) * 5
Y_data = poly_bai11(X_data)
# Chẻ dữ liệu đầu vào thành [x^2, x, 1]
XX_data = np.hstack([
    X_data * X_data,
    X_data,
    np.ones_like(X_data)
])
# AI khởi tạo 3 hệ số ngẫu nhiên
w_bai11 = torch.randn(3, 1, requires_grad=True)
x_tensor = torch.tensor(XX_data, dtype=torch.float32)
y_tensor = torch.tensor(Y_data, dtype=torch.float32)
optimizer11 = torch.optim.NAdam([w_bai11], lr=0.01)
# Vòng lặp học tập
for epoch in range(1500): # Tăng lên 1500 vòng để AI ra kết quả mịn hơn
    optimizer11.zero_grad()
    y_predict = x_tensor @ w_bai11
    loss_mse = torch.mean(torch.square(y_tensor - y_predict))
    loss_mse.backward()
    optimizer11.step()
print("- Phương trình gốc có hệ số chuẩn là: [2, -3, 5]")
print("- Ba hệ số AI đã học được là:\n", w_bai11.detach().numpy())
print("=> Nhận xét: Các hệ số học được đã xấp xỉ vô cùng sát với [2.0, -3.0, 5.0]. Mô hình huấn luyện rất thành công!")
```
