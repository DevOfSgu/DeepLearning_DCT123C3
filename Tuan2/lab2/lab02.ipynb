## Bài 1. Tạo Tensor cơ bản
Tạo một Tensor số nguyên kích thước $2 \times 3$ và hiển thị kết quả.
Yêu cầu: Cho biết kích thước, số chiều và kiểu dữ liệu của Tensor a.

```python
import torch

a = torch.tensor(
    [[1, 2, 3],
     [4, 5, 6]],
    dtype=torch.int32
)

print("Tensor a:\n", a)
print("-----------------")

# Trả lời yêu cầu:
print("Kích thước (shape):", a.shape)
print("Số chiều (ndim):", a.ndim)
print("Kiểu dữ liệu (dtype):", a.dtype)
```

## Bài 2. Tạo các Tensor thường dùng
Thực hành tạo Tensor ngẫu nhiên, Tensor toàn số 0, toàn số 1 và ma trận đơn vị.

**Yêu cầu:**
* Nhận xét sự khác nhau giữa `torch.rand()` và `torch.randn()`.
* Tạo thêm một Tensor $5 \times 5$ chứa toàn giá trị 10.

```python
import torch

a = torch.rand(3, 4)
b = torch.randn(3, 4)
c = torch.zeros(2, 3)
d = torch.ones(2, 3)
e = torch.eye(4)

print("a (rand):\n", a)
print("\nb (randn):\n", b)
print("\nc (zeros):\n", c)
print("\nd (ones):\n", d)
print("\ne (eye):\n", e)

print("\n--- Trả lời yêu cầu ---")
# Nhận xét: 
# - torch.rand() sinh ra các giá trị ngẫu nhiên phân bố đều (uniform distribution) nằm trong khoảng từ 0 đến 1.
# - torch.randn() sinh ra các giá trị ngẫu nhiên theo phân bố chuẩn (normal/Gaussian distribution), có trung bình bằng 0 và độ lệch chuẩn bằng 1 (nên có cả số âm).

# Tạo Tensor 5x5 chứa toàn giá trị 10
# Cách 1:
f1 = torch.ones(5, 5) * 10
# Cách 2 (chuyên nghiệp hơn):
f2 = torch.full((5, 5), 10.0)

print("\nTensor 5x5 chứa toàn 10 (Cách 2):\n", f2)
```
